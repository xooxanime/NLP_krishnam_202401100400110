```python
import pandas as pd
import re
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
df = pd.read_csv(
    '/content/IMDB Dataset.csv',
    engine='python',
    on_bad_lines='skip'
)

print(df.shape)
df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
print(df['sentiment'].value_counts())
```

```python
# removing <tags>
def remove_html(text):
    return re.sub(r'<.*?>', ' ', text)

df['cleaned_review'] = df['review'].apply(remove_html)
```

```python
print("Original:")
print(df['review'].iloc[1])

print("\nAfter HTML removal:")
print(df['cleaned_review'].iloc[1])
```

```python
#lower casing
df['cleaned_review'] = df['cleaned_review'].str.lower()
```

```python
print(df['cleaned_review'].iloc[1])
```

```python
#Punctuation & special characters
df['cleaned_review'] = df['cleaned_review'].apply(
    lambda x: re.sub(r'[^a-zA-Z\s]', ' ', x)
)
```

```python
#Space removal and compression
df['cleaned_review'] = df['cleaned_review'].str.replace(r'\s+', ' ', regex=True).str.strip()
```

```python
print(df['cleaned_review'].iloc[1])
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
df['tokens'] = df['cleaned_review'].apply(word_tokenize)
```

```python
print(df['tokens'].iloc[1])
```

```python
stop_words = set(stopwords.words('english'))

df['tokens'] = df['tokens'].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)
```

```python
print(df['tokens'].iloc[1])
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
df['tokens'] = df['tokens'].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)
```

```python
print(df['tokens'].iloc[1])
```

```python
#Rejoin
df['cleaned_review'] = df['tokens'].apply(
    lambda tokens: ' '.join(tokens)
)
```
